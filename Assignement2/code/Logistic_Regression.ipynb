
## 3. Sigmoid Function

Logistic Regression uses the **sigmoid** function:

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

This maps any real number to (0,1).


```python

def sigmoid(z):
    # TODO: Implement sigmoid
    #implementing sigmoid function by using formula
    return 1 / (1 + np.exp(-z))
    pass

```


## 4. Model Prediction

The logistic regression model is:

\[
z = Xw + b
\]
\[
\hat{y} = \sigma(z)
\]

where:
- `w` → weight vector
- `b` → bias


```python

def predict_proba(X, w, b):
    # TODO: Compute z and apply sigmoid

    # @ operator is used for matrix multiplication, so using that here
    z = X @ w + b

    # Step 2: Appling sigmoid function using the previous cell formula
    y_hat = sigmoid(z)

    return y_hat
    # Hint : Use @ (search for it if you don't know)
    pass
```
